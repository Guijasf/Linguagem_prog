## 3. Tratamento e preparação dos dados

```python
# Conversão da data para datetime
df["DataPedido"] = pd.to_datetime(df["DataPedido"])

# Garantindo que Preço e Unidades sejam numéricos
df["Preço"] = pd.to_numeric(df["Preço"], errors="coerce")
df["Unidades"] = pd.to_numeric(df["Unidades"], errors="coerce")

# Verificando os tipos após a conversão
df.dtypes
```

```python
print("Valores ausentes por coluna:")
display(df.isna().sum())

print("\nQuantidade de linhas duplicadas:", df.duplicated().sum())
```

## 4. Criando novas colunas

A coluna `TotalVenda` representa o valor total do pedido:

**TotalVenda = Preço × Unidades**

Também criaremos as colunas `Ano` e `Mes` para facilitar análises temporais.

```python
df["TotalVenda"] = df["Preço"] * df["Unidades"]
df["Ano"] = df["DataPedido"].dt.year
df["Mes"] = df["DataPedido"].dt.month

df[["DataPedido", "Vendedor", "Item", "Preço", "Unidades", "TotalVenda", "Ano", "Mes"]].head()
```

## 5. Filtros

```python
# Pedidos realizados no Rio de Janeiro
pedidos_rj = df[df["UF"] == "RJ"]
pedidos_rj.head()
```

```python
# Pedidos com mais de 50 unidades
mais_50_unidades = df[df["Unidades"] > 50]
mais_50_unidades[["DataPedido", "Estado", "Vendedor", "Item", "Unidades"]].head(10)
```

## 6. Agrupamentos com groupby

```python
# Faturamento total por estado
faturamento_estado = (
    df.groupby("Estado", as_index=False)["TotalVenda"]
      .sum()
      .sort_values("TotalVenda", ascending=False)
)

faturamento_estado
```

```python
# Quantidade total de unidades vendidas por vendedor
unidades_vendedor = (
    df.groupby("Vendedor", as_index=False)["Unidades"]
      .sum()
      .sort_values("Unidades", ascending=False)
)

unidades_vendedor
```

```python
# Faturamento por seção
faturamento_secao = (
    df.groupby("Seção", as_index=False)["TotalVenda"]
      .sum()
      .sort_values("TotalVenda", ascending=False)
)

faturamento_secao
```

```python
# Resumo por item
resumo_item = (
    df.groupby("Item")
      .agg(
          Quantidade_Pedidos=("Item", "count"),
          Unidades_Vendidas=("Unidades", "sum"),
          Preco_Medio=("Preço", "mean"),
          Faturamento=("TotalVenda", "sum")
      )
      .sort_values("Faturamento", ascending=False)
)

resumo_item
```

## 7. Ordenação: maiores pedidos

```python
top_10_pedidos = (
    df.sort_values("TotalVenda", ascending=False)
      [["DataPedido", "Estado", "Vendedor", "Item", "Preço", "Unidades", "TotalVenda"]]
      .head(10)
)

top_10_pedidos
```

## 8. Tabela dinâmica

```python
tabela_dinamica = pd.pivot_table(
    df,
    values="TotalVenda",
    index="Vendedor",
    columns="Ano",
    aggfunc="sum",
    fill_value=0
)

tabela_dinamica
```

## 9. Gráficos

```python
# Faturamento por estado
grafico_estado = faturamento_estado.set_index("Estado")["TotalVenda"]

ax = grafico_estado.plot(kind="bar", figsize=(8, 4))
ax.set_title("Faturamento total por estado")
ax.set_xlabel("Estado")
ax.set_ylabel("Faturamento (R$)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Faturamento por ano
faturamento_ano = df.groupby("Ano")["TotalVenda"].sum()

ax = faturamento_ano.plot(kind="line", marker="o", figsize=(8, 4))
ax.set_title("Faturamento por ano")
ax.set_xlabel("Ano")
ax.set_ylabel("Faturamento (R$)")
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Total de unidades por vendedor
grafico_vendedores = unidades_vendedor.set_index("Vendedor")["Unidades"]

ax = grafico_vendedores.plot(kind="bar", figsize=(8, 4))
ax.set_title("Unidades vendidas por vendedor")
ax.set_xlabel("Vendedor")
ax.set_ylabel("Unidades")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
